# append
1. appends an element to the end of a list:
2. you can enter one element at a time
3. you can append both iterable and non iterable object
See Function can have only one element append(b) that's why append can enter one element at a time

```python
a = [1,2,3]
a.append(4)
print(a)
```

<!-- extend -->
It is similar to concatenation but it does not create the new list object rather it updates into the existing object.
You can use only iterable objects


```python
a = [1,2,3]
b = [5,6,7]
# a+b  This is concatenation

a.extend(b)
print(a)
# This will add individual element in the a

a.append(b)
print(a)
#  This will add whole list of b directly
# [1, 2, 3, [5, 6, 7]]
```

<!-- Insert -->
Insert method insert the value at a particular index
- only one element can be inserted in a list at a time

Syntax: List_1.insert(index, element to be inserted)

```python
a = [10,20,30]
b = 100
a.insert(2,b)
print(a)
```

```python
a = [1,2,3]
b = [1,2,[4,5],10]
a.extend(b)
print(a)
```

```python
a = [10,20,30]
b = 100
a.insert(-3,b)
print(a)
```
