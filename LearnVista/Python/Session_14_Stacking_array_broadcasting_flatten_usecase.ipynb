## Vector Stacking

Concatenating of two arrays is called vector stacking

```python
import numpy as np
```

```python
a = np.array([["New York", "Marryland"], ["Albany","annapolis"]])
b = np.array([["Ohio", "Arizona"], ["columbus","phoenix"]])
```

```python
a
```

```python
b
```

```python
np.concatenate((a,b)) # axix = 0 row wise
```

```python
np.concatenate((a,b), axis=1) # axix = 1 column wise
```

## Keep in mind things 
1. When axis = 0 then column of both the matrix should be same
2. when axix = 1 then row of both the matrix should be same

```python
a = np.array([["New York", "Marryland"], ["Albany","annapolis"],["goa","banglore"]])
b = np.array([["Ohio", "Arizona"], ["columbus","phoenix"]])
```

```python
a.shape
```

```python
a
```

```python
b.shape
```

```python
np.concatenate((a,b), axis = 0)
```

```python
# alternative method instead of axis

np.hstack((a,b))
```

```python
np.vstack((a,b))
```

ArrayBroadCasting

```python
c = np.arange(6).reshape(2,3)
c
```

```python
d = np.arange(6,12).reshape(2,3)
d
```

```python
print(c+d)
```

```python
# What if different shape

e = np.arange(6).reshape(2,3)
f = np.arange(3).reshape(1,3)
```

```python
print(e)
print(f)
```

```python
print(e + f)
```
