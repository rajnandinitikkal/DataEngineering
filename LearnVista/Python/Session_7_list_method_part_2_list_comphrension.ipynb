<!-- Remove elements from list -->
1. pop
2. remove

<!--  Pop -->
Removes the element by index
pop() by default removes the last element
It is only method that returns the value
Syntax: l1.pop(index)

```python
l1 = [10,20,30]
l1.pop(1)
# print(l1)
```

```python
l1 = [10,20,30]
l1.pop()
# By default removes last element
```

<!-- Remove -->
Removes the specify value from the list
remove() method removes the first occurrence of a value

```python
a = [10,20,30,20,30]
a.remove(20)
print(a)
```

<!-- Count -->
count() takes in an element and return the number of times it occures in your list:

```python
l2 = [10,0, 20, 10, 10, 40, 50,50]
l2.count(10)
```

<!-- Index -->
Returns the index of the element. Make a note that if the element is not in the list then it returns an error.

```python
l2 = [10,0, 20, 10, 10, 40, 50,50]
l2.index(20)
```

<!-- Reverse -->
As the name suggests, reverse() helps you to reverse a list. Note this occurs in place! Meaning it effects your list permanently.


```python
l = [1,2,3,4,5]
l.reverse()
l
```

<!-- Sort -->
Sort will sort your list in place:
Note: sort will only work on list which are having homogeneous data (homogeneus: Not mixture of other than one datatype it will be only one type like, numeric, string or similar kind of data)

```python
a = [1,4,3,4,35,7,8,3,7,10,33,56,79]
a.sort()
a
```

```python
# This will sort data in descending order
a = [1,4,3,4,35,7,8,3,7,10,33,56,79]
a.sort(reverse=True)
a
```

# Tuple: 
Tuple is similar to list
All the element should be kept inside a round (optional) and seprated by comas

```python
a = (1,2,3,4)
print(a, type(a))
```

# Properties od tuple

1. Sequence data type
2. Orderd collection of the element -> the element are indexed
3. supports +ve and -ve indexing
4. supports indexing and slicing
    Indexing - accessing one element
    Slicing - accessing multiple element at a time
5. immutable
6. Allows duplicate element
7. Allows to store different object types or data types like string, tuple, set , dictionary, int, float atc.
8. Basic operations - Slicing, indexing, concatination, repetation, Membership

```python
# 5. Immutable
# Cannot modify tuple

a = (10, 20, 30)
a[0] = 100
```

```python
# 6. Allows duplicate element
a = (10,10,10)
print(a, type(a))
```

```python
# 7. Allows to store different object types or data types like string, tuple, set , dictionary, int, float atc.

a = (1, 2, 4.5 ,[5,6], "Rutuja")
a
```

```python
# Slicing
a = (1, 2, 4.5 ,[5,6], "Rutuja")
a[-1]
```

```python
a = (1, 2, 4.5 ,[5,6], "Rutuja")
a[::-1]
```

```python
# Concatenation

a = (1,2,3)
b = (5,6,7)
a + b
```

```python
# Repetation
a = (1,2,3)
b = 5
a * b
```

```python
# Identity

a = (1,2,3)
b = (1,2,3)

a is b
```
