# Anonymous function(function without a name)- lambda function
- A lambda function is a small anonymous function
- A lambda function can take any number of arguments, but can only have one expression
- Ayntax: lambda arguments: expression

```python
# Two ways of calling lambda function
# 1.
(lambda a : a%2 == 0)(24)

# 2.
name = lambda a,b : a+b 
name(10,30)
```

```python
# add a number using function
(lambda a : a%2 == 0)(24)
```

```python
is_even = lambda a: "even" if a%2 == 0 else "odd"
# is_even(10)
is_even(21)
```
